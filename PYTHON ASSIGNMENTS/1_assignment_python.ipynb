Q6. Create a list and access the first, last, and third elements.
list_values : ["Apple", "Banana", "Mango", "Orange", "Grapes"]
- Add "Pineapple" to the list.
- Insert "Watermelon" at index 1.
- Remove "Banana" from the list.


```python
list_values = ["Apple", "Banana", "Mango", "Orange", "Grapes"]
print(list_values[0])
print(list_values[-1])
print(list_values[2])
list_values.append("Pineapple")
print(list_values)
list_values.insert(1,"Watermelon")
print(list_values)
print(list_values.remove("Banana"))
print(list_values)
```

Q10. Create a dictionary for a student. (name, age, city)
- Access the student's name.
- Add a new key "salary".
- Update the age.
- Remove the city key.
- Display all keys and values separately.


```python
student={
         "Name":"Indu",
         "age":21,
         "city":"Visakhapatnam"
}
print(student["Name"])
student["salary"]=50000
print(student)
student["age"]=22
print(student)
student.pop("city")
print(student)
print(student.keys())
print(student.values())
```

Q11. Merge two lists.
