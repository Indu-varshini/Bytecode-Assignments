- Find the average (marks = [70, 80, 90, 60])

```python
marks = [70, 80, 90, 60]
total = 0
for i in marks:
  total = total+ i
  avg = total/len(marks)
print(total)
print(avg)
```

- Calculate total sales

```python
sales={"shirts":100,
       "jeans":3000,
       "tops":200,
       "shorts":400}
total_sales=0
for i in sales.values():
  total_sales=total_sales+i
print(total_sales)
```

- Find highest sales value

```python
sales={"shirts":100,
       "jeans":3000,
       "tops":200,
       "shorts":400}
for i in sales.values():
  largest=max(sales.values())
print(largest)
```

- Count customers with purchases above 1000

```python
sales={"indu":100,
       "pavi":3000,
       "likki":2000,
       "usha":400
       }
count=0
for i in sales.values():
  if i>1000:
    count=count+1
print(count)
```
